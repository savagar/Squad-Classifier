```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing.sequence import pad_sequences
from scipy.interpolate import interp1d
from sklearn.preprocessing import MinMaxScaler
import joblib

def extract_squat_sequences(df, label_mapping):
    """
    - df: IMU 센서 데이터 (Pandas DataFrame)
    - label_mapping: { (Button1, Button2): Label } 형태의 딕셔너리
    """
    squat_sequences = []
    labels = []
    current_squat = []
    last_button_state = None 

    for i, row in df.iterrows():
        button_1 = row["Button1"]
        button_2 = row["Button2"]

        if (button_1, button_2) in label_mapping: 
            current_squat.append(row.iloc[:-2].values) 
            last_button_state = (button_1, button_2) 

        elif button_1 == 0 and button_2 == 0 and last_button_state is not None:
            final_label = label_mapping.get(last_button_state, None)
            if final_label is not None:
                squat_sequences.append(current_squat)
                labels.append(final_label)

            current_squat = []
            last_button_state = None

    return squat_sequences, labels

def resample_sequence(sequence, target_length=50):
    current_length = len(sequence)
    
    if current_length < 2:
        return np.tile(sequence, (target_length, 1))
    
    if current_length == target_length:
        return sequence
    
    new_time_index = np.linspace(0, current_length - 1, target_length)
    
    # Interpolation
    interpolated_sequence = np.array([
        interp1d(np.arange(current_length), sequence[:, i], kind="linear", fill_value="extrapolate")(new_time_index)
        for i in range(sequence.shape[1])
    ]).T  
    
    return interpolated_sequence

def normalize_sequences(sequences):

    scaler = MinMaxScaler()  
    all_data = np.vstack(sequences) 
    scaler.fit(all_data)  
    normalized_sequences = [scaler.transform(seq) for seq in sequences]

    return normalized_sequences, scaler

# 1 : simple, 2: narrow, 3:plie
label_mapping_df1 = {(1,0): 2, (1,1): 3}
label_mapping_df2 = {(1,0): 3}
label_mapping_df3 = {(1,1): 3}
label_mapping_df4 = {(1,1): 3}
label_mapping_df5 = {(1,1): 3}
label_mapping_df6 = {(1,0): 3}
label_mapping_df7 = {(0,1): 1}
label_mapping_df8 = {(1,0): 1}
label_mapping_df9 = {(0,1): 1}
label_mapping_df10 = {(0,1): 1}
label_mapping_df11 = {(0,1): 1}
label_mapping_df17 = {(0,1): 1}
label_mapping_df18 = {(1,0): 1}
label_mapping_df12 = {(1,0): 2}
label_mapping_df13 = {(1,0): 2}
label_mapping_df14 = {(0,1): 2}
label_mapping_df15 = {(1,0): 2}
label_mapping_df16 = {(1,0): 2}
label_mapping_df19 = {(0,1): 2, (1,0): 1, (1,1): 3}
label_mapping_df20 = {(0,1): 1, (1,0): 2, (1,1): 3}
label_mapping_df21 = {(0,1): 1, (1,0): 2, (1,1): 3}
label_mapping_df22 = {(0,1): 2, (1,0): 1}
label_mapping_df23 = {(0,1): 2, (1,0): 1}


squat_sequences_1, labels_1 = extract_squat_sequences(df1, label_mapping_df1)
squat_sequences_2, labels_2 = extract_squat_sequences(df2, label_mapping_df2)
squat_sequences_3, labels_3 = extract_squat_sequences(df3, label_mapping_df3)
squat_sequences_4, labels_4 = extract_squat_sequences(df4, label_mapping_df4)
squat_sequences_5, labels_5 = extract_squat_sequences(df5, label_mapping_df5)
squat_sequences_6, labels_6 = extract_squat_sequences(df6, label_mapping_df6)
squat_sequences_7, labels_7 = extract_squat_sequences(df7, label_mapping_df7)
squat_sequences_8, labels_8 = extract_squat_sequences(df8, label_mapping_df8)
squat_sequences_9, labels_9 = extract_squat_sequences(df9, label_mapping_df9)
squat_sequences_10, labels_10 = extract_squat_sequences(df10, label_mapping_df10)
squat_sequences_11, labels_11 = extract_squat_sequences(df11, label_mapping_df11)
squat_sequences_17, labels_17 = extract_squat_sequences(df17, label_mapping_df17)
squat_sequences_18, labels_18 = extract_squat_sequences(df18, label_mapping_df18)
squat_sequences_12, labels_12 = extract_squat_sequences(df12, label_mapping_df12)
squat_sequences_13, labels_13 = extract_squat_sequences(df13, label_mapping_df13)
squat_sequences_14, labels_14 = extract_squat_sequences(df14, label_mapping_df14)
squat_sequences_15, labels_15 = extract_squat_sequences(df15, label_mapping_df15)
squat_sequences_16, labels_16 = extract_squat_sequences(df16, label_mapping_df16)
squat_sequences_19, labels_19 = extract_squat_sequences(df19, label_mapping_df19)
squat_sequences_20, labels_20 = extract_squat_sequences(df20, label_mapping_df20)
squat_sequences_21, labels_21 = extract_squat_sequences(df21, label_mapping_df21)
squat_sequences_22, labels_22 = extract_squat_sequences(df22, label_mapping_df22)
squat_sequences_23, labels_23 = extract_squat_sequences(df23, label_mapping_df23)
squat_sequences = squat_sequences_1 + squat_sequences_2 + squat_sequences_3 + squat_sequences_4 + squat_sequences_5 + squat_sequences_6 + squat_sequences_7 + squat_sequences_8 + squat_sequences_9 + squat_sequences_10 + squat_sequences_11 + squat_sequences_12 + squat_sequences_13 + squat_sequences_14 + squat_sequences_15 + squat_sequences_16 + squat_sequences_17 + squat_sequences_18 + squat_sequences_19 + squat_sequences_20 + squat_sequences_21 + squat_sequences_22 + squat_sequences_23
labels = labels_1 + labels_2 + labels_3 + labels_4 + labels_5 + labels_6 + labels_7 + labels_8 + labels_9 + labels_10 + labels_11 + labels_12 + labels_13 + labels_14 + labels_15 + labels_16 + labels_17 + labels_18 + labels_19 + labels_20 + labels_21 + labels_22 + labels_23

print(f"✅ 기존 데이터 스쿼트 개수: {len(squat_sequences_1)}")
print(f"✅ 새로운 데이터 스쿼트 개수: {len(squat_sequences_23)}")
print(f"✅ 전체 스쿼트 개수: {len(squat_sequences)}")

unique_labels, counts = np.unique(labels, return_counts=True)

print("\n📊 라벨별 스쿼트 샘플 개수:")
for label, count in zip(unique_labels, counts):
    print(f"🔹 Label {label} (Squat form {label}) Sample Number: {count}")
```

```python
# Padding
y = np.array(labels)

target_length = 50
squat_sequences = [resample_sequence(np.array(seq), target_length) for seq in squat_sequences]

squat_sequences, scaler = normalize_sequences(squat_sequences)
joblib.dump(scaler, "fitted_scaler1.pkl")

X = pad_sequences(squat_sequences, maxlen=target_length, dtype="float32", padding="post", truncating="post")
X_train, X_val, y_train, y_val = train_test_split(X, labels, test_size=0.2, random_state=42, stratify=labels)

print("✅ 데이터 로드 및 처리 완료!")
print(f"📊 X 데이터 형태: {X.shape}")  
print(f"📊 y 데이터 형태: {y.shape}")  
print(f"📊 X_train.shape: {X_train.shape}")  
print(f"📊 X_val.shape: {X_val.shape}") 

print("✅ 학습 데이터 준비 완료!")


print(X_train[0])
```

```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense, Dropout, LayerNormalization, MultiHeadAttention, GlobalAveragePooling1D, BatchNormalization
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping
import numpy as np

def transformer_encoder(inputs, head_size, num_heads, ff_dim, dropout=0.1):
    # Multi-Head Self Attention
    attention = MultiHeadAttention(num_heads=num_heads, key_dim=head_size, dropout=dropout)(inputs, inputs)
    attention = Dropout(dropout)(attention)
    attention = LayerNormalization(epsilon=1e-6)(inputs + attention)
    # Feed Forward Network (FFN)
    ffn = Dense(ff_dim, activation="relu", kernel_regularizer=l2(0.01))(attention)
    ffn = Dropout(dropout)(ffn)
    ffn = Dense(inputs.shape[-1], kernel_regularizer=l2(0.01))(ffn)
    # Batch Normalization
    ffn = BatchNormalization()(ffn)
    ffn = LayerNormalization(epsilon=1e-6)(attention + ffn)  # Residual Connection + LayerNorm
    return ffn

def build_transformer_model(input_shape, num_classes, head_size=128, num_heads=8, ff_dim=256, num_transformer_blocks=4, dropout=0.1):
    inputs = Input(shape=input_shape)

    x = inputs
    for _ in range(num_transformer_blocks):
        x = transformer_encoder(x, head_size, num_heads, ff_dim, dropout)

    # Global Average Pooling
    x = GlobalAveragePooling1D()(x)
    outputs = Dense(num_classes, activation="softmax", kernel_regularizer=l2(0.01))(x)
    model = Model(inputs, outputs)
    return model

batch_size = 16 
epochs = 200  

y_train = np.array(y_train) - 1  # {1, 2, 3} → {0, 1, 2}
y_val = np.array(y_val) - 1

num_classes = len(set(y_train))  # 3

model = build_transformer_model(input_shape=(50, X_train.shape[2]), num_classes=num_classes)
model.compile(loss="sparse_categorical_crossentropy", optimizer=Adam(learning_rate=0.0005), metrics=["accuracy"])

reduce_lr = ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=5, min_lr=1e-6, verbose=1)
early_stopping = EarlyStopping(monitor="val_loss", patience=10, restore_best_weights=True)

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=epochs,
    batch_size=batch_size,
    callbacks=[reduce_lr],
    verbose=1
)

loss, acc = model.evaluate(X_val, y_val)
print(f"Validation Accuracy: {acc:.4f}")
```
